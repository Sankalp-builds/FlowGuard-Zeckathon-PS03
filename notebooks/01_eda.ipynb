Notebook companion to the reproducible pipeline. Run from the project root or this notebook folder.

```python
from pathlib import Path
import sys
import pandas as pd
import matplotlib.pyplot as plt
ROOT=Path.cwd() if (Path.cwd()/'data').exists() else Path.cwd().parent
sys.path.insert(0,str(ROOT/'src'))
from data_pipeline import load_dataset, quality_report, TARGETS
df=load_dataset(ROOT/'data/processed/monthly_dataset.csv')
print(df.shape, df.month.iloc[0], df.month.iloc[-1])
quality_report(df)
```

```python
display(df[TARGETS].describe().T)
display(quality_report(df).query('missing_count > 0'))
```

```python
fig,axes=plt.subplots(2,1,figsize=(12,7),sharex=True)
for ax,target in zip(axes,TARGETS):
    ax.plot(pd.to_datetime(df.month),df[target],marker='o'); ax.axhline(0,color='black',lw=.7); ax.set_title(target+' (positive=inflow, negative=outflow)')
plt.tight_layout()
```

```python
corr=df.select_dtypes('number').corr()
plt.figure(figsize=(12,9)); plt.imshow(corr,vmin=-1,vmax=1,cmap='coolwarm'); plt.colorbar(); plt.xticks(range(len(corr)),corr.columns,rotation=90,fontsize=7); plt.yticks(range(len(corr)),corr.columns,fontsize=7); plt.tight_layout()
```
