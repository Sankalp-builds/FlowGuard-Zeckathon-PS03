# Out-of-sample driver analysis

Driver value is measured by changes in rolling-origin error when feature groups are added to the autoregressive ML model. The comparisons are predictive, not causal.

```python
from pathlib import Path
import sys
import pandas as pd
ROOT=Path.cwd() if (Path.cwd()/'data').exists() else Path.cwd().parent
sys.path.insert(0,str(ROOT/'src'))
from data_pipeline import load_dataset
from evaluation import driver_ablation
df=load_dataset(ROOT/'data/processed/monthly_dataset.csv')
metrics,fold_predictions=driver_ablation(df)
metrics.sort_values(['target','horizon','MAE_reduction_vs_base_pct'],ascending=[True,True,False])
```

```python
# Negative improvement means the group increased held-out forecast error.
display(metrics)
print('Saved pipeline artifacts:',ROOT/'outputs/driver_oos_comparison.csv')
```
